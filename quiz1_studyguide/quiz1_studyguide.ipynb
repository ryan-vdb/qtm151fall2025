#### 1. Use an if/elif/else block to test whether a string belongs to one of two lists. Depending on the outcome (belongs to list A, belongs to list B, doesn't belong to either list) print an appropriate message to the screen.

```python
item = "pear"

list_a = ["apple", "banana", "pear"]
list_b = ["carrot", "celery", "lettuce"]

if item in list_a:
    print(f"{item!r} belongs to list A")
elif item in list_b:
    print(f"{item!r} belongs to list B")
else:
    print(f"{item!r} belongs to neither list")
```

#### 2. You are given two lists of numbers, x_vals and y_vals, and must create a Numpy array where the components are related to those in the lists (for example, you might be asked to create a Numpy array with entries). Then generate a histogram of the values in the Numpy array, with labeled axes and a title. You will also be asked to change the color of the bars in the histogram.

```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = [1, 2, 3, 4, 5]
y_vals = [10, 9, 8, 7, 6]

# Example transformation: a_i = x_i^2 + y_i
a = np.array([x**2 + y for x, y in zip(x_vals, y_vals)])

plt.figure()
plt.hist(a, bins=10, color="orange")  # change color here
plt.xlabel("Transformed values")
plt.ylabel("Frequency")
plt.title("Histogram of x^2 + y")
plt.show()
```

#### 3. Generate random samples from two specified distributions and plot their histograms, side by side in a single figure. Each sub-figure will have its own labeled axes and title.

```python
import numpy as np
import matplotlib.pyplot as plt

n = 1000
sample1 = np.random.normal(loc=0, scale=1, size=n)     # Normal(0,1)
sample2 = np.random.exponential(scale=1.0, size=n)     # Exponential(λ=1)

fig, axes = plt.subplots(1, 2, figsize=(10, 4), tight_layout=True)

axes[0].hist(sample1, bins=30)
axes[0].set_title("Normal(0,1)")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Frequency")

axes[1].hist(sample2, bins=30)
axes[1].set_title("Exponential(1)")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Frequency")

plt.show()
```

#### 4. Read in a .csv file to create a Pandas DataFrame. Create a scatter plot with one of the variables (or features) corresponding to the x-axis and another corresponding to the y-axis. The plot should have a title and labeled axes.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")  # replace with your filename
x_col = "feature_x"
y_col = "feature_y"

plt.figure()
plt.scatter(df[x_col], df[y_col])
plt.title(f"{y_col} vs {x_col}")
plt.xlabel(x_col)
plt.ylabel(y_col)
plt.show()
```

#### 5. Read in a .csv file to create a Pandas DataFrame. Loop over a list of variable/column names, printing the mean of each variable.

```python
import pandas as pd

df = pd.read_csv("data.csv")
cols = ["feature_x","feature_y","feature_z"]  # numeric columns

for c in cols:
    print(c + "mean: " + str(df[c].mean()))
```

#### 6. Create a list of .csv file names and an additional empty list. Then in a for loop, read in each file to make a DataFrame, compute the mean of one of the variables/features/columns, and store it as an element of the originally empty list. Then print the list.

```python
import pandas as pd

files = ["a.csv", "b.csv", "c.csv"]
target_col = "col_of_interest"
means = []

for f in files:
    df = pd.read_csv(f)
    means.append(df[target_col].mean())

print("Means:", means)
```
